```python
import pathlib
import polars as pl
import plotly.express as px
import plotly.graph_objects as go
```

```python
## Config

INSURANCE_DATA_PATH = pathlib.Path("../data/raw/insurance_data")
OUTPUT_PATH = pathlib.Path("../data/interim/insurance_data/combined_insurance_data.txt")

```

```python
file_pattern = "*.txt"
order = sorted(INSURANCE_DATA_PATH.glob(file_pattern))
print(order)
```

```python
def combine_files(files, output_path):
    """Combine multiple text files into one output file.
    
    Args:
        files: List of Path objects for input files
        output_path: Path object for output file
    """
    try:
        # Create output directory if it doesn't exist
        output_path.parent.mkdir(parents=True, exist_ok=True)
        
        with output_path.open("w", encoding="utf-8", newline="\n") as outfile:
            for file in files:
                try:
                    with file.open("r", encoding="utf-8") as src:
                        content = src.read()
                        if content:  # Only write if content is not empty
                            outfile.write(content + "\n")  # Add newline between files
                        print(f"Successfully processed {file.name}")
                except Exception as e:
                    print(f"Error processing {file.name}: {str(e)}")
                    continue
        
        print(f"\nCombined file created at: {output_path}")
        print(f"Output file size: {output_path.stat().st_size} bytes")
        
    except Exception as e:
        print(f"Error creating combined file: {str(e)}")
```

```python
combine_files(order, OUTPUT_PATH)
```

```python
insurance_data = pl.read_csv(OUTPUT_PATH, 
                             separator="|")

fields = [
    "commodity_year",
    "state_code",
    "state_abbreviation",
    "county_code",
    "county_name",
    "commodity_code",
    "commodity_name",
    "insurance_plan_code",
    "insurance_plan_abbreviation",
    "coverage_category",
    "delivery_type",
    "coverage_level",
    "policies_sold_count",
    "policies_earning_premium_count",
    "policies_indemnified_count",
    "units_earning_premium_count",
    "units_indemnified_count",
    "quantity_type",
    "net_reported_quantity",
    "endorsed_companion_acres",
    "liability_amount",
    "total_premium_amount",
    "subsidy_amount",
    "state_private_subsidy",
    "additional_subsidy",
    "efa_premium_discount",
    "indemnity_amount",
    "loss_ratio"
]


insurance_data.columns = fields
```

```python
## Checking for state counts

state_counts = insurance_data["state_abbreviation"].value_counts().sort("count", descending=True)
state_counts

```

```python

fig = px.bar(state_counts.to_pandas(),
             x="state_abbreviation",
             y="count",
             title="Number of Insurance Policies by State",
             labels={"state_abbreviation": "State", "count": "Number of Policies"})
fig.show()
```

```python
## combining the county name and state abbreviation for better visualization

insurance_data = insurance_data.with_columns(
    (pl.col("county_name").str.strip_chars() + "," + pl.col("state_abbreviation").str.strip_chars()).alias("county_state")
)
insurance_data.tail()

```

```python

## Plotting the top 20 counties with the highest number of insurance policies
top_20_counties = insurance_data["county_state"].value_counts().sort("count", descending=True).head(20)
fig = px.bar(top_20_counties.to_pandas(), x="county_state", y="count", title="Top 20 Counties by Insurance Policies",
             labels={"county_state": "County, State", "count": "Number of Policies"})
fig.show()
```

```python
## Years with most insurance policies
year_counts = insurance_data["commodity_year"].value_counts().sort("count", descending=True)
year_counts
```

```python
## Visualizing the number of insurance policies by year
fig = px.bar(year_counts, x="commodity_year", y="count",
             title="Number of Insurance Policies by Year",
             labels={"commodity_year": "Year", "count": "Number of Policies"})
fig.show()
```

```python
## Checking the indemnity amounts by state
state_indemnity = insurance_data.group_by("state_abbreviation").agg(
    pl.sum("indemnity_amount").alias("total_indemnity")
).sort("total_indemnity", descending=True)
state_indemnity
```

```python
## Visualizing the number of insurance loss by state

fig = px.bar(state_indemnity.to_pandas(), x="state_abbreviation", y="total_indemnity",
             title="Total Indemnity Amount by State",
             labels={"state_abbreviation": "State", "total_indemnity": "Total Indemnity"})
fig.show()
```

```python
## Checking indemnity amounts by counties
county_indemnity = insurance_data.group_by("county_state").agg(
    pl.sum("indemnity_amount").alias("total_indemnity")
).sort("total_indemnity", descending=True)  

## Getting the top 20 counties with the highest indemnity amounts
county_indemnity = county_indemnity.head(20)
county_indemnity
```

```python
## Visualizing the indemnity amounts by counties

fig = px.bar(county_indemnity.to_pandas(), x="county_state", y="total_indemnity",
             title="Total Indemnity Amount by County",
             labels={"county_state": "County", "total_indemnity": "Total Indemnity"})

fig.show()
```

```python
## Checking counties with highest loss ratios
county_loss_ratio = insurance_data.group_by("county_state").agg(
    pl.mean("loss_ratio").alias("loss_ratio")
).sort("loss_ratio", descending=True)
county_loss_ratio
```

```python
## Visualizing the loss ratios by counties

fig = px.bar(county_loss_ratio.head(20).to_pandas(), x="county_state", y="loss_ratio",
             title="Top 20 Counties by Average Loss Ratio",
             labels={"county_state": "County, State", "loss_ratio": "Average Loss Ratio"})
fig.show()
```
